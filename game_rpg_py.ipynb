```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name # Nama Hero
        self.hp = hp # Nyawa (Health Point)
        self.attack_power = attack_power # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 10)
balmond = Hero("Balmond", 200, 10)

eudora.info()
#eudora.serang(balmond) # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
```

```python
from abc import ABC, abstractmethod

class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = 0  # Private attribute untuk stok
        self.__harga_dasar = harga_dasar # Private attribute untuk harga dasar

    # Abstract method
    @abstractmethod
    def tampilkan_detail(self):
        pass

    # Abstract method
    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

    # Getter untuk stok
    def get_stok(self):
        return self.__stok

    # Setter/Method untuk menambah stok dengan validasi
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Jumlah tidak boleh negatif ({jumlah}).")
            return False
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {jumlah} unit. Total stok: {self.__stok}")
            return True

    # Getter untuk harga dasar (meskipun private, kita beri getter jika perlu)
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Getter untuk nama kamar
    def get_nama_kamar(self):
        return self.nama_kamar
```

```python
class KamarDeluxe(KamarHotel):
    PAJAK_RATE = 0.10 # 10%

    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        print(f"[DELUXE] {self.get_nama_kamar()} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak({int(self.PAJAK_RATE*100)}%): Rp {self.get_harga_dasar() * self.PAJAK_RATE:,.0f}")

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam_plus_pajak = self.get_harga_dasar() * (1 + self.PAJAK_RATE)
        return harga_per_malam_plus_pajak * jumlah_malam

class KamarStandard(KamarHotel):
    PAJAK_RATE = 0.05 # 5%

    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        print(f"[STANDARD] {self.get_nama_kamar()} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak({int(self.PAJAK_RATE*100)}%): Rp {self.get_harga_dasar() * self.PAJAK_RATE:,.0f}")

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam_plus_pajak = self.get_harga_dasar() * (1 + self.PAJAK_RATE)
        return harga_per_malam_plus_pajak * jumlah_malam
```

```python
def proses_transaksi(daftar_pesanan):
    total_tagihan_akhir = 0
    print("--- STRUK PEMESANAN ---")
    for i, (kamar, jumlah_malam) in enumerate(daftar_pesanan):
        subtotal = kamar.hitung_harga_total(jumlah_malam)
        print(f"{i+1}. ", end="")
        kamar.tampilkan_detail()
        print(f"Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}")
        total_tagihan_akhir += subtotal
    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan_akhir:,.0f}")
    print("----------------------------------------")
```

```python
# a) Admin membuat data kamar
kamar_deluxe_sea_view = KamarDeluxe("Kamar Deluxe Sea View", 1500000, "Private Pool")
kamar_standard_superior = KamarStandard("Kamar Standard Superior", 500000, "2 Orang")

print("--- SETUP DATA KAMAR ---")
# b) Admin mencoba mengisi stok kamar dengan angka negatif
kamar_standard_superior.tambah_stok(-5)

# Admin mengisi stok kamar dengan angka positif
kamar_deluxe_sea_view.tambah_stok(10)
kamar_standard_superior.tambah_stok(20)

# c) Tamu memesan
daftar_pesanan_tamu = [
    (kamar_deluxe_sea_view, 2), # Memesan Kamar Deluxe 2 malam
    (kamar_standard_superior, 1) # Memesan Kamar Standard 1 malam
]

# d) Program menampilkan detail kamar yang dipesan dan total tagihan akhir
proses_transaksi(daftar_pesanan_tamu)
```
